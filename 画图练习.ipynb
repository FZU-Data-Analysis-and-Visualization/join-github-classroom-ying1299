```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.family"] = ["SimHei", "Microsoft YaHei"]

fy = ["FY2018", "FY2019", "FY2020", "FY2021", "FY2022"]
brandA = [15970, 18770, 22020, 26890, 31150]
brandB = [8470, 12150, 15650, 21310, 26120]

x = np.arange(len(fy))
width = 0.35
plt.figure(figsize=(11,7), dpi=120)

rect_a = plt.bar(x - width/2, brandA, width, color='#2266bb', edgecolor='white', alpha=0.95)
rect_b = plt.bar(x + width/2, brandB, width, color="#22c4ff", edgecolor='white', alpha=0.95)

#添加顶部数字标签
def autolabel(rects):
    for rect in rects:
        rect_height = rect.get_height()
        rect_x = rect.get_x()
        rect_width = rect.get_width()
        plt.text(
            rect_x + rect_width / 2,
            rect_height + 150,
            s=f'{rect_height:,}',
            fontsize=9, weight='medium'
        )
#顶部数值
autolabel(rect_a)
autolabel(rect_b)

# 图表标题与坐标轴
plt.title("2018-2022财年品牌A、B GMV规模对比（单位：亿元）", fontsize=13, pad=25)
plt.ylabel("GMV（亿元）", fontsize=10)
plt.xticks(x, fy, fontsize=9)
plt.yticks(fontsize=9)
plt.ylim(0, 33000)

plt.legend([rect_a, rect_b], ['品牌A', '品牌B'], loc='upper center', ncol=2, bbox_to_anchor=(0.5, 1.00))

plt.subplots_adjust(top=0.82)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.family"] = ["SimHei", "Microsoft YaHei"]

fy = ["FY2018", "FY2019", "FY2020", "FY2021", "FY2022"]
brandA = [15970, 18770, 22020, 26890, 31150]
brandB = [8470, 12150, 15650, 21310, 26120]

x = np.arange(len(fy))
width = 0.35
plt.figure(figsize=(11,7), dpi=120)

rect_a = plt.bar(x - width/2, brandA, width, color='#2266bb', edgecolor='white', alpha=0.95)
rect_b = plt.bar(x + width/2, brandB, width, color="#22c4ff", edgecolor='white', alpha=0.95)

# 添加顶部数字标签
def autolabel(rects):
    for rect in rects:
        rect_height = rect.get_height()
        rect_x = rect.get_x()
        rect_width = rect.get_width()
        plt.text(
            rect_x + rect_width / 2,
            rect_height + 150,
            s=f'{rect_height:,}',
            ha='center', va='bottom',
            fontsize=9, weight='medium'
        )
#顶部数值
autolabel(rect_a)
autolabel(rect_b)


#差值
diff_data = [brandA[i] - brandB[i] for i in range(len(fy))]
plt.plot(x, diff_data, color='#009933', marker='^', linewidth=2, label='A-B GMV差值')
for xi, val in zip(x, diff_data):
    plt.text(xi, val + 300, f'{val:,}', ha='center', fontsize=8, color='#009933')


# 网格
plt.grid(axis='y', alpha=0.3, linestyle='--')

# 图表标题与坐标轴
plt.title("2018-2022财年品牌A、B GMV规模对比（单位：亿元）", fontsize=13, pad=25)
plt.ylabel("GMV（亿元）", fontsize=10)
plt.xticks(x, fy, fontsize=9)
plt.yticks(fontsize=9)
plt.ylim(0, 33000)

plt.legend([rect_a, rect_b], ['品牌A', '品牌B'], loc='upper left')

plt.subplots_adjust(top=0.82)
plt.tight_layout()
plt.show()
```
